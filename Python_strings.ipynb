#3. String slicing

```python
print(stg1)
```

```python
print(stg1[0:5])
```

```python
stg1[1:5] # it gives the string elements from index 1to5
```

```python
stg1[5:] # it slise the string from index5
```

```python
stg1[0::] # it gives the string as it is
```

```python
stg1[::-1] # it reverse the string elements
```

```python
stg1[0:20:2] # here 0 is starting index 20 is ending index and 2 step
```

# 4.String methods

```python
print(stg1)
```

```python
print(stg1.upper()) # the entire string is printed in upper cases
```

```python
print(stg1.lower()) # the entire string is printed in lower cases
```

```python
print(stg1.find('m')) # It returns the index value of first occurrence on the value
```

```python
print (stg1.index('m')) # It work as same as find method
```

```python
print(stg1.split()) # It returns the string as list
```

```python
print(stg1.split(" "))  # It returns the string as list
```

```python
stg1.replace('my world','my life') # It replace the existing value with given value (first parameter is existing value),(second parameter is new value )
```

```python
print (stg1.rpartition(' to ')) # It returns a string as tuple
```
