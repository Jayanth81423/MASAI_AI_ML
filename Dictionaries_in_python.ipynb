# Dictionaries
Dictionariy is an unordered collection of data stored as a pair of key and value

# Creating a dictionary

```python
d1={}   # Empty dictionary
print (d1)
print (type(d1))
```

```python
d2 = {1:'welcome',2:'to',3:'my',4:'world'}
print(d2)
```

```python
d3={'name':'jayant','age':22,'profession':'student'}
print(d3)
```

```python
d4=dict({1:'welcome',2:'to',3:'my',4:'world'})
print(d4) # By using dict keyword
```

```python
d5=dict([(1,'welcome'),(2,'to'),(3,'my'),(4,'world')])  # Creating a dictionary with lists
print(d5)
```

# Nested dictionary

```python
d6={'name':{'first':'jayant','second':'katherashala'},'age':22,'profession':'student'}
print(d6)
```

# Adding elements to a dictionary

```python
d={}
d[0]='welcome'
print(d)
```

```python
d[1]=('How' , 'are' , 'you')
print(d)
```

```python
print(d[1]) # To print the element in specific index
```
