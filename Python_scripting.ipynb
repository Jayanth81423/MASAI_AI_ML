# Nested functions

```python
def func1():
    x=10
    def func2(x):
        return x+1
    return func2(x)
result=func1()
print(result)
```

```python
def function1(called_function):
    print("this is the first function")
    def nested_function1(called_function):
        print ("This is the nested function")
        called_function()  # Call the passed function
    return nested_function1(called_function)
def outer_function():
    print("This is the outer function")


obj=function1(outer_function)
```

# Creating Class at run time is known as factory

```python
#Factory
B=type("BaseClass", (object,), {})
c1=type("C1",(B,),{'val':5})
c2=type('c2',(B,),{'val':10})


def ClassCreator(bool):
    if bool:
        return c1
    else:
        return c2

print(ClassCreator(True).val)
print(ClassCreator(False).val)
```
