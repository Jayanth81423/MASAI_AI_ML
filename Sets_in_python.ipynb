# Set
set is an unordered collection of unique elements

```python
s= set([1,2,3,4])
print(s)
print(type(s))
```

```python
s.add('a')
print(s)
```

```python
fs=frozenset([1,2,3,4])
print(fs)
```

```python
fs.add('a')
```

# Mathamatical oparations on set's

```python
s1=set([1,5,3,6,2,6,7,8,9])
s2=set([1,6,3,8,2,8,92,0,1])
```

```python
s1.union(s2)
```

```python
s1.difference(s2)
```
