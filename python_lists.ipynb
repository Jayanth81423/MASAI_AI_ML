# 1. creating a list of numbers

```python
num=[3,3,5,2,6,5,4,6]
print(num)
```

```python
letter= ['a','s','h','h','d','d','f']
print(letter)
```

```python
stg= ["I'm",'jayanth','katherashala','of','20','years','old']
print (stg)
print(type(stg))
```
