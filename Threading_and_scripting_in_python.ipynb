# Thread
* A thread is squence of instructions in a program that can be executed independently of the remaining program

```python
from threading import *
def show ():
    print("This is a child thread")
t= Thread(target = show())
t.start()
print ('this is parent thread')
```

```python
from threading import *
class Mythread (Thread):
    def run(self):
        for i in range (20):
            print("\nThis is a child clsass")
t= Mythread()
t.start()
for i in range (5):
    print("\nThis is the main thread")
```

```python
from threading import *
class Demo:
    def show (self):
        for i in range (5):
            print("\nThis is a child thread")
obj= Demo()
t= Thread(target=obj.show())
t.start()
for i in range(5):
    print ("\nThis is the parent thread")
```

# Multithreading
* A model where multiple threds within a process execute independently while sharing the same resources is called multithreading

```python
from threading import *
import time
class Demo:
    def num (self):
        for i in range (1,6):
            print("The number is",i)
            time.sleep(1)
    def double (self):
        for i in range (1,6):
            print("The double is",2*i)
            time.sleep(1)
    def square (self):
        for i in range (1,6):
            print("The square is",i*i)
            time.sleep(1)
obj= Demo()
t1= Thread(target=obj.num)
t2= Thread(target=obj.double)
t3= Thread(target=obj.square)
t1.start()
time.sleep(0.2)
t2.start()
time.sleep(0.2)
t3.start()
time.sleep(0.2)
t1.join()
t2.join()
t3.join()

```
